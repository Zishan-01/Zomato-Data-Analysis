```python
import pandas as pd
import sqlite3

df = pd.read_csv('../data/zomato_cleaned.csv')

conn = sqlite3.connect('../sql/zomato.db')
df.to_sql('restaurants', conn, if_exists='replace', index=False)

cursor = conn.cursor()
cursor.execute("SELECT COUNT(*) FROM restaurants")
print(f"Rows in DB: {cursor.fetchone()[0]:,}")
conn.close()
print("✅ Database created: sql/zomato.db")
```
