```python
import pandas as pd

# Load the raw dataset
df = pd.read_csv('../data/zomato.csv')

print(f"Rows    : {df.shape[0]}")
print(f"Columns : {df.shape[1]}")
print("\nColumn names:")
print(df.columns.tolist())

df.head()
```

```python
# Basic shape
print(f"Rows    : {df.shape[0]:,}")
print(f"Columns : {df.shape[1]}")
print("\nColumns:")
for i, col in enumerate(df.columns, 1):
    print(f"{i:2}. {col}")

print("\nData Types:")
print(df.dtypes)
```

```python
# Missing values audit
missing = pd.DataFrame({
    'Missing Count': df.isnull().sum(),
    'Missing %': (df.isnull().sum() / len(df) * 100).round(2)
}).sort_values('Missing %', ascending=False)

missing[missing['Missing Count'] > 0]
```

```python
print(f"Duplicate rows: {df.duplicated().sum()}")

print("\nUnique values in key columns:")
for col in ['name', 'location', 'rest_type', 'cuisines', 'listed_in(city)']:
    print(f"{col:22} → {df[col].nunique():,} unique")
```

```python
# Drop columns with no analytical value
cols_to_drop = ['url', 'phone', 'menu_item', 'reviews_list', 'dish_liked']
df = df.drop(columns=cols_to_drop)

print(f"Remaining columns: {df.shape[1]}")
print(list(df.columns))
```

```python
# Replace 'NEW' and '-' with NaN
df['rate'] = df['rate'].replace(['NEW', '-', 'NEW '], np.nan)

# Remove '/5' suffix
df['rate'] = df['rate'].astype(str).str.replace('/5', '', regex=False)

# Convert to float
df['rate'] = pd.to_numeric(df['rate'], errors='coerce')

# Verify
print(df['rate'].describe())
print(f"Nulls in rate: {df['rate'].isnull().sum()}")
```

```python
# Remove commas
df['approx_cost(for two people)'] = (
    df['approx_cost(for two people)']
    .astype(str)
    .str.replace(',', '', regex=False)
)

# Convert to numeric
df['approx_cost(for two people)'] = pd.to_numeric(
    df['approx_cost(for two people)'], errors='coerce'
)

# Rename for convenience
df = df.rename(columns={'approx_cost(for two people)': 'cost_for_two'})

print(df['cost_for_two'].describe())
```

```python
# Keep only valid Yes/No
df['online_order'] = df['online_order'].where(df['online_order'].isin({'Yes', 'No'}))
df['book_table']   = df['book_table'].where(df['book_table'].isin({'Yes', 'No'}))

# Drop rows where our main KPI (rate) is missing
before = len(df)
df = df.dropna(subset=['rate'])

# Fill categorical missing with 'Unknown'
for col in ['location', 'rest_type', 'cuisines']:
    df[col] = df[col].fillna('Unknown')

# Fill numeric missing
df['cost_for_two'] = df['cost_for_two'].fillna(df['cost_for_two'].median())
df['votes'] = df['votes'].fillna(0).astype(int)
df['cost_for_two'] = df['cost_for_two'].astype(int)

# Drop duplicates
before_dup = len(df)
df = df.drop_duplicates(subset=['name', 'address', 'location', 'cuisines'])

print(f"Rows dropped (missing rate): {before - before_dup:,}")
print(f"Duplicate rows removed: {before_dup - len(df):,}")
print(f"Final row count: {len(df):,}")
```

```python
print("=" * 50)
print("CLEANED DATA SUMMARY")
print("=" * 50)
print(f"Rows           : {df.shape[0]:,}")
print(f"Columns        : {df.shape[1]}")
print(f"Unique rest.   : {df['name'].nunique():,}")
print(f"Avg rating     : {df['rate'].mean():.2f}")
print(f"Median cost    : ₹{df['cost_for_two'].median():,.0f}")
print(f"Online order % : {(df['online_order'] == 'Yes').mean()*100:.1f}%")
print(f"Book table %   : {(df['book_table'] == 'Yes').mean()*100:.1f}%")
print("=" * 50)

# Save the cleaned dataset to the data folder
df.to_csv('../data/zomato_cleaned.csv', index=False)
print("✅ Saved: ../data/zomato_cleaned.csv")
```
